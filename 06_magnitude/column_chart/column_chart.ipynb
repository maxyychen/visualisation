# Column Chart (vertical bar)

**Family:** Magnitude &nbsp;|&nbsp; **Library:** matplotlib / seaborn &nbsp;|&nbsp; **Level:** Basic

> *Magnitude:* Compare sizes, relative or absolute (usually counts rather than rates).

## Design purpose

Vertical bars whose heights encode values on a common zero baseline. Aligned length is the most accurately read visual encoding, which makes this the default for comparing amounts.

## When to use

- Compare values across a small number (≤ 12) of categories with short labels
- Show counts or amounts per time period (months, quarters)

## Avoid when

- Category labels are long (use horizontal bars)
- The baseline is not zero (bar length would lie)
- The data is a continuous trend with many points (use a line)

## Alternatives

- Horizontal bar chart
- Dot plot
- Lollipop chart

## Design tips

- Always start the y-axis at zero
- One series means one color. Highlight a single bar only if it is the story
- Gaps between bars ~30–50% of the bar width

## Example

Quarterly revenue for two years (illustrative data), highlighting the latest quarter.

### Setup

```python
%matplotlib inline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Library style: categorical colors in fixed order, recessive axes, hairline grid.
PALETTE = ["#2a78d6", "#eb6834", "#1baf7a", "#eda100",
           "#e87ba4", "#008300", "#6250d6", "#e34948"]
SEQ_CMAP = "Blues"      # sequential: one hue, light -> dark
DIV_CMAP = "RdBu_r"     # diverging: two opposite hues, neutral midpoint
GRAY = "#b8b7b1"        # de-emphasised marks
plt.rcParams.update({
    "figure.figsize": (8, 4.8), "figure.dpi": 110, "figure.facecolor": "#fcfcfb",
    "axes.facecolor": "#fcfcfb", "axes.edgecolor": "#c9c8c3", "axes.linewidth": 0.8,
    "axes.spines.top": False, "axes.spines.right": False,
    "axes.grid": True, "grid.color": "#e6e5e1", "grid.linewidth": 0.6, "axes.axisbelow": True,
    "axes.prop_cycle": plt.cycler(color=PALETTE),
    "axes.titlesize": 13, "axes.titleweight": "bold", "axes.titlelocation": "left",
    "axes.labelcolor": "#52514e", "xtick.color": "#52514e", "ytick.color": "#52514e",
    "text.color": "#0b0b0b", "font.size": 10, "lines.linewidth": 2,
    "legend.frameon": False,
})
```

### Data

```python
df = pd.DataFrame({
    "quarter": ["Q1 23", "Q2 23", "Q3 23", "Q4 23", "Q1 24", "Q2 24", "Q3 24", "Q4 24"],
    "revenue_m": [4.2, 4.8, 5.1, 6.3, 5.0, 5.6, 6.1, 7.4],
})
df
```

### Plot

```python
fig, ax = plt.subplots(figsize=(8, 4.5))
colors = [PALETTE[0] if i == len(df) - 1 else "#9ec5f4" for i in range(len(df))]
bars = ax.bar(df["quarter"], df["revenue_m"], color=colors, width=0.65)
ax.bar_label(bars, fmt="$%.1fM", padding=3, fontsize=9, color="#52514e")
ax.set_ylabel("Revenue ($M)")
ax.set_ylim(0, df["revenue_m"].max() * 1.15)
ax.grid(axis="x", visible=False)
ax.set_title("Quarterly revenue: Q4 2024 is a record")
fig.tight_layout()
plt.show()
```

## How to read it

Each bar is one quarter's revenue, all measured from zero. Revenue climbs through each year and peaks in Q4 both times ($6.3M, then $7.4M); the highlighted Q4 2024 is the highest of all eight quarters.
